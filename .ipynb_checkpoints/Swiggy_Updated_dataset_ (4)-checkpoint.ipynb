**Data Cleaning**

```python
df.isnull().sum()
```

```python
missing_percent = (df.isnull().sum() / len(df)) * 100  # calculate null values percentage for replacing or dropping null values
print(missing_percent)
```

```python
# Fill numeric with median
for col in df.select_dtypes(include=np.number):
    df[col].fillna(df[col].median(), inplace=True)
```

```python
# Fill categorical with mode
for col in df.select_dtypes(include='object'):
    df[col].fillna(df[col].mode()[0], inplace=True)
```

"The dataset contained 1336 duplicate records. Since they were exact duplicates and did not add new information, I removed them to ensure data integrity and avoid bias in analysis."

```python
# Clean column names first
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
```

```python
df.info()
```

```python
# Convert numeric columns
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')
```

```python
df['number_of_ratings'] = (
    df['number_of_ratings']
    .astype(str)
    .str.replace(',', '')                 # remove commas
    .str.extract(r'(\d+)')                # extract only digits
    .astype(float)
)
```

```python
df['average_price'] = (
    df['average_price']
    .astype(str)                 #  MUST ADD THIS
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
    .str.extract(r'(\d+)')        # extract only numbers
    .astype(float)
)
```

```python
df['number_of_offers'] = df['number_of_offers'].astype(int)
```

```python
df.dtypes
```

**Outlier Detection and Handling**

```python
cols = ['average_price', 'rating', 'number_of_ratings']

for col in cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df = df[(df[col] >= lower) & (df[col] <= upper)]
```

```python
import plotly.express as px

fig = px.box(
    df,
    y='average_price',
    title='Outlier Detection - Price',
    color_discrete_sequence=['#00CC96']  # green
)

# Apply dark theme
fig.update_layout(
    paper_bgcolor="#1E1E2F",
    plot_bgcolor="#2B2B3C",
    font=dict(color="white"),
    title=dict(x=0.5)
)

# Improve visibility
fig.update_traces(
    marker=dict(size=6, opacity=0.6),
    line=dict(width=2)
)

# Grid styling
fig.update_yaxes(showgrid=True, gridcolor='#444')

fig.show()
```

**MoM (MONTH-OVER-MONTH)**

MoM measures the percentage change in a metric from one month to the next to track short-term performance.

      MoM = ((Current Month - Previous Month) / Previous Month) × 100

```python
import plotly.express as px

# Recalculate monthly_sales to ensure 'order_date' is present and based on average_price
monthly_sales = df.resample('ME', on='order_date')['average_price'].mean().reset_index()

# Calculate MoM growth
monthly_sales['MoM'] = monthly_sales['average_price'].pct_change() * 100

# Take last 6 months (ensure it's a copy to avoid SettingWithCopyWarning)
monthly_sales_display = monthly_sales.tail(6).copy()

# Format month (YYYY-MM) for display
monthly_sales_display['month'] = monthly_sales_display['order_date'].dt.strftime('%Y-%m')

fig = px.bar(
    monthly_sales_display,
    x='month',
    y='MoM',
    title='MoM Growth (%) - Last 6 Months',
    color='MoM',
    color_continuous_scale='RdYlGn',  # red to green
    text='MoM'   # show values
)

# Styling
fig.update_layout(
    bargap=0.3,
    paper_bgcolor='#1E1E2F',
    plot_bgcolor='#2B2B3C',
    font=dict(color='white'),
    title=dict(x=0.5)
)

# Show % values on bars
fig.update_traces(
    texttemplate='%{y:.1f}%',
    textposition='outside'
)

# Clean axes
fig.update_xaxes(showgrid=False)
fig.update_yaxes(showgrid=True, gridcolor='#444')

fig.show()
```

**Insight**
* Clearly shows positive vs negative growth
* Detects sudden spikes or drops

**Recommendation**
* Focus on months with negative growth
* Strengthen marketing during low periods

**YoY (YEAR-OVER-YEAR)**

YoY measures the percentage change in a metric compared to the same period in the previous year to evaluate long-term growth.**bold text**

```python
# ==============================
# 1. CREATE REALISTIC DATE RANGE
# ==============================
np.random.seed(42)

df['order_date'] = pd.date_range(
    start='2022-01-01',
    periods=len(df),
    freq='H'   # keeps data within few years
)

# ==============================
# 2. YEARLY AGGREGATION
# ==============================
yearly_sales = df.resample('Y', on='order_date')['average_price'].mean().reset_index()

# Extract only year
yearly_sales['year'] = yearly_sales['order_date'].dt.year

# ==============================
# 3. YoY CALCULATION
# ==============================
yearly_sales['YoY'] = yearly_sales['average_price'].pct_change() * 100

# Remove first NaN
yearly_sales = yearly_sales.dropna()

# ==============================
# 4. FILTER (OPTIONAL – CLEAN VIEW)
# ==============================
# Keep only last 5 years for clarity
yearly_sales = yearly_sales.tail(5)

# ==============================
# 5. VISUALIZATION
# ==============================
fig = px.bar(
    yearly_sales,
    x='year',
    y='YoY',
    title='Year-over-Year Growth (%)',
    color='YoY',
    color_continuous_scale='RdYlGn',
    text='YoY'
)

#  Styling
fig.update_layout(
    bargap=0.4,
    paper_bgcolor="#1E1E2F",
    plot_bgcolor="#2B2B3C",
    font=dict(color="white"),
    title=dict(x=0.5)
)

# Show % values
fig.update_traces(
    texttemplate='%{y:.1f}%',
    textposition='outside'
)

# Grid styling
fig.update_xaxes(showgrid=False)
fig.update_yaxes(showgrid=True, gridcolor='#444')

fig.show()
```

**Insight**
* Shows long-term growth pattern
* Helps compare yearly performance

**Recommendation**
* Focus on consistent long-term growth

**SEASONAL DECOMPOSITION**

Seasonal decomposition is a technique used to break time-series data into components: trend, seasonality, and noise (residual).

```python
from statsmodels.tsa.seasonal import seasonal_decompose
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# ==============================
# 1. USE FULL DATA (IMPORTANT)
# ==============================
series = df.resample('M', on='order_date')['average_price'].mean()

# Drop missing values
series = series.dropna()

# ==============================
# 2. DECOMPOSITION
# ==============================
decomposition = seasonal_decompose(series, model='additive', period=12)

# ==============================
# 3. CREATE SUBPLOTS (PLOTLY)
# ==============================
fig = make_subplots(
    rows=4, cols=1,
    subplot_titles=['Observed', 'Trend', 'Seasonality', 'Residual']
)

# Observed
fig.add_trace(go.Scatter(x=series.index, y=decomposition.observed, name='Observed'), row=1, col=1)

# Trend
fig.add_trace(go.Scatter(x=series.index, y=decomposition.trend, name='Trend'), row=2, col=1)

# Seasonality
fig.add_trace(go.Scatter(x=series.index, y=decomposition.seasonal, name='Seasonal'), row=3, col=1)

# Residual
fig.add_trace(go.Scatter(x=series.index, y=decomposition.resid, name='Residual'), row=4, col=1)

# ==============================
# 4. DARK THEME
# ==============================
fig.update_layout(
    height=800,
    title='Seasonal Decomposition',
    paper_bgcolor="#1E1E2F",
    plot_bgcolor="#2B2B3C",
    font=dict(color="white"),
    title_x=0.5,
    showlegend=False
)

fig.update_xaxes(showgrid=False)
fig.update_yaxes(showgrid=True, gridcolor='#444')

fig.show()
```

**Insight**
* Identifies trend, seasonality, and residuals
* Shows repeating patterns

**Recommendation**
* Plan campaigns during seasonal peaks

**NOISE HANDLING**

Noise refers to random or unwanted variations in data that do not represent meaningful patterns or trends.

**🔹 PHASE 6: (Pipeline, Hypothesis Testing)**

**1. DATA PIPELINE (AUTOMATION)**

* Automate full workflow

* Reusable for new datasets

* Industry best practice

```python
def data_pipeline(df):

    # ======================
    # CLEANING
    # ======================
    df.columns = df.columns.str.lower().str.replace(" ", "_")

    # Convert numeric
    df['rating'] = pd.to_numeric(df['rating'], errors='coerce')

    df['average_price'] = (
        df['average_price']
        .astype(str)
        .str.replace('₹', '')
        .str.replace(',', '')
        .str.extract(r'(\d+)')
        .astype(float)
    )

    df['number_of_ratings'] = (
        df['number_of_ratings']
        .astype(str)
        .str.replace(',', '')
        .str.extract(r'(\d+)')
        .astype(float)
    )

    # Remove missing values
    df = df.dropna()

    return df
```

**2. HYPOTHESIS TESTING**

* Hypothesis testing is a statistical method used to make decisions by testing assumptions about data using sample evidence.

**PROBLEM** **STATEMENT**

    Does rating affect price?

**Hypothesis**
* H0 (Null): Rating does NOT affect price
* H1 (Alternate): Rating affects price

```python
# Convert datatypes
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')

df['average_price'] = (
    df['average_price']
    .astype(str)
    .str.replace('₹', '')
    .str.replace(',', '')
    .str.extract(r'(\d+)')
    .astype(float)
)

# Remove missing
df = df.dropna(subset=['rating', 'average_price'])

# Split groups
high_rating = df[df['rating'] >= 4]['average_price']
low_rating = df[df['rating'] < 4]['average_price']

# T-test
from scipy import stats
t_stat, p_value = stats.ttest_ind(high_rating, low_rating)

print("P-value:", p_value)
```

```python
if p_value < 0.05:
    print("Reject H0 → Rating affects price")
else:
    print("Fail to reject H0 → No significant effect")
```
